## Task 1 — clean_key(text)

A key is a cleaned-up version of a name that we only use for comparing.

**What to do**

Write clean_key(text). It returns text in lowercase, with no spaces at the start or end and only one space between words.

```
clean_key("     SINDH")              gives   "sindh"
clean_key("Kambar  Shahdad Kot ")    gives   "kambar shahdad kot"
```

**Useful tools**

```
.lower()          "SINDH".lower() gives "sindh"
.split()          "  Kambar   Shahdad Kot ".split()
                  gives ["Kambar", "Shahdad", "Kot"]
" ".join(words)   " ".join(["Kambar", "Shahdad", "Kot"])
                  gives "Kambar Shahdad Kot"
```

````python
def clean_key(text):
    """Return a lowercase version of text with tidy single spaces."""
    # >>> TODO (Task 1): write your code here <<<
    lower = text.lower()
    lower = lower.split()
    for i in range(len(lower)):
        if lower[i] == ' ':
            lower.remove(lower[i])
    return " ".join(lower)
````

### Task 2 — count_records(records)

**What to do**

Write count_records(records). It returns a dictionary that gives, for the clean_key of each district, how many records it has. Same counting pattern as count_kinds in Session 1.

```
count_records(records)   gives   {"khairpur": 8930, "dadu": 7739, ...}
```

````python
def count_records(records):
    """Return a dictionary: cleaned district name -> number of records."""
    m = {}
    for r in records:
        x = clean_key(r["district"])
        if x not in m:
            m[x] = 1
        else:
            m[x] += 1

    return m
````

## Given code — count_people(records)

Another way to measure help: the people_reached column. count_people works like your count_records, but adds each record's people_reached instead of 1. (It skips the 3,699 records where people_reached is empty.)

````python
def count_people(records):
    """GIVEN: return a dictionary: cleaned district name -> total people reached."""
    totals = {}
    for r in records:
        if r["people_reached"] == "":
            continue
        key = clean_key(r["district"])
        people = float(r["people_reached"])
        if key in totals:
            totals[key] = totals[key] + people
        else:
            totals[key] = people
    return totals


banner("SECTION 3 — People reached")

people_counts = count_people(records)
if people_counts is None:
    print("count_people isn't written yet.")
else:
    print("People reached in Dadu:", round(people_counts.get("dadu", 0)))
    print()
    print("People reached, added up over ALL records:", round(sum(people_counts.values())))
    print("People the government said were affected:  about 33,000,000")
````

## Task 3 — weekly_peak(row)

A fairer "hit hardest" uses only the 8 weekly windows.

**What to do**

Write weekly_peak(row). row is one row of district_peak.csv. It returns the most people exposed (a float) in any window in WEEKLY_LABELS.

**Useful tools**

row["exposed " + label] gives that window's value, as a string. max(a_list) gives the biggest value in a list.

````python
WEEKLY_LABELS = WINDOW_LABELS[1:]      # everything except "01-31 Aug"
def weekly_peak(row):
    """Return the most people exposed in any weekly window, as a float."""
    # >>> TODO (Task 3): write your code here <<<
    highest = 0
    for label in WEEKLY_LABELS:
        value = float(row["exposed " + label])
        if value > highest:
            highest = value
    return highest
````

## Section 5 — Help per province: does it match the need?

Now we compare whole provinces. Given code: build_province_tree builds the tree from Session 2 again (Pakistan -> provinces -> districts), and each district holds three numbers:

```
"exposed"   its weekly peak         (the need)
"records"   its relief records      (help, one way)
"people"    its people reached      (help, another way)
```

````python
banner("SECTION 5a — Building the province tree")

def build_province_tree(districts, record_counts, people_counts):
    """GIVEN: return the top place (Pakistan) with provinces and districts inside."""
    top = {"name": "Pakistan", "level": "country", "children": []}
    current_province = None

    for row in districts:
        # 1. A new province? Make it, and put it inside Pakistan.
        if current_province is None or row["province"] != current_province["name"]:
            current_province = {"name": row["province"], "level": "province",
                                "children": []}
            top["children"].append(current_province)

        # 2. Make the district, and put it inside the current province.
        key = clean_key(row["district"])
        district = {
            "name": row["district"],
            "level": "district",
            "exposed": weekly_peak(row),
            "records": record_counts.get(key, 0),
            "people": people_counts.get(key, 0),
        }
        current_province["children"].append(district)

    return top


relief_tree = None
if ready_a:
    relief_tree = build_province_tree(districts, record_counts, people_counts)
    for province in relief_tree["children"]:
        print("  ", province["name"], "-", len(province["children"]), "districts")
else:
    print("Finish Tasks 1-3 first, then run this again.")
````

**Given code** — tree_total(place, key): the same recursion as your total_population in Session 2, for any of the three numbers.

````python
def tree_total(place, key):
    """GIVEN: return the total of place[key] over all districts inside place."""
    if place["level"] == "district":
        return place[key]
    total = 0
    for child in place["children"]:
        total = total + tree_total(child, key)
    return total


banner("SECTION 5b — Each province's share of the need, and of the help")

if relief_tree is None:
    print("Finish Tasks 1-3 first, then run this again.")
else:
    whole = {}
    for key in ["exposed", "records", "people"]:
        whole[key] = tree_total(relief_tree, key)

    print("   province              % of people   % of relief   % of people")
    print("                           exposed        records        reached")
    for province in relief_tree["children"]:
        print("  ", province["name"].ljust(20), end="")
        for key in ["exposed", "records", "people"]:
            share = 100 * tree_total(province, key) / whole[key]
            print(str(round(share)).rjust(12), end="  ")
        print()
````

**A choice we just made:** we judge "fair" by comparing each province's share of the help with its share of the need (people exposed at the weekly peak).
